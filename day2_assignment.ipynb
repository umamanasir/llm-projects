```python
OLLAMA_BASE_URL = "http://localhost:11434/v1"
OLLAMA_MODEL = "llama3.2"


ollama = OpenAI(base_url = OLLAMA_BASE_URL, api_key="ollama")

```

```python
ed = fetch_website_contents("https://edwarddonner.com")
```

```python
system_prompt = """
You are an experienced website summarizer. Create a humorous summary of the website that you are given. Ignore navigation related text and respond in markdown. 
"""

user_prompt = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too: 

"""
```

```python
def message_summary(website):
        return [
            {"role": "system", "content": system_prompt}, 
            {"role": "user", "content": user_prompt + website}
        ]
```

```python
def ollama_call(website): 
    message = message_summary(website)
    print(message)
    print("Passing msg to Ollama")
    response = ollama.chat.completions.create(model = OLLAMA_MODEL, messages = message)
    print("Response generated")
    return response.choices[0].message.content
```

```python
def summarize_websites(website): 
    print("Calling Ollama")
    response = ollama_call(website)
    return response
    
```

```python
summary = summarize_websites("https://edwarddonner.com")
summary
```
