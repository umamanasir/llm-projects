```python
import os 
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI

```

```python
load_dotenv(override =True)
api_key = os.getenv('GOOGLE_API_KEY')

if not api_key:
    print("API key not available")
elif api_key: 
   print("API key available and loaded.")
```

```python
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
gemini = OpenAI(base_url = GEMINI_BASE_URL, api_key=api_key)
MODEL_NAME = 'gemini-3.1-flash-lite'
```

```python
system_prompt = """You are an advanced project manager who tracks project delivery, timeline, budget and scopes. 
You support an existing project manager in their tasks, create GANTT charts for them and 
provide all necessary support with regards to documentation etc.  
"""
```

```python
def messages_pm(task):
        return [{"role": "system", "content": system_prompt}, 
                    {"role": "user", "content": task }]
```

```python
def model_pm(task):
    response = gemini.chat.completions.create(
        model = MODEL_NAME, 
        messages = messages_pm(task)
    )
    return response.choices[0].message.content

```

```python
task = """
I'm the project manager for a mid-sized software company rolling out a new internal CRM system. The project has 12 team members across engineering, QA, and design. We're currently in week 6 of a 16-week timeline. Two key issues have come up:

The backend API integration is running about 2 weeks behind schedule due to unexpected complexity in syncing with the legacy sales database.
Our QA lead just informed me she's going on leave for 3 weeks starting next Monday, and we don't have a clear backup plan for test coverage during that time.
We report progress to executive stakeholders every two weeks, and our next steering committee meeting is in 4 days. Budget is fixed and cannot be increased. The client (an internal sales department) is expecting a soft launch by week 14 to allow for 2 weeks of user acceptance testing before the week 16 go-live date.

Task:

Draft a status update for the steering committee that:
Honestly communicates the delay and staffing gap
Proposes 2-3 concrete mitigation options with trade-offs (e.g., timeline, scope, resourcing)
Recommends one path forward with reasoning
Keeps the tone professional but not alarmist

"""
```

```python
response = model_pm(task)
display(Markdown(response))
```
